1.Use a list comprehension to build a list of the squares of the numbers 1 through 10. 
  

```python
ans=[x**2 for x in range(1,11)]
print(ans)
```

2.  Use a list comprehension to extract only the vowels from a given string. 

```python
word="OmkarSanjayPatil"

ans=[char for char in word if char in ["a","e","i","o","u"]]
print(ans)
```

3.  Use a dictionary comprehension to build a dict that maps the numbers 1 through 5 to their cubes

```python
dic={num:num**3 for num in range(1,6)}
print(dic)
```

4.  Use a list comprehension with a conditional expression to create a list labeling each number from 1 
to 10 as 'even' or 'odd'. 

```python
ans=["Even" if num%2==0 else "Odd" for num in range(1,11)]
print(ans)
```

5.  Given a list of words, use a dictionary comprehension to map each word to its length.

```python
words = ["Python", "Java", "C++", "JavaScript"]

ans={word:len(word) for word in words}
print(ans)
```

6.  Given a 3x3 matrix represented as a list of lists, use a single nested list comprehension to flatten it 
into one flat list. 

```python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

flat=[ num for row in matrix for num in row]
print(flat)
```

7.  Use a dictionary comprehension to invert a dictionary, but handle the case where multiple keys share 
the same value: instead of overwriting, each value should map to a list of all keys that had it. 

```python
##flagged
data={
    'a':1,
    'b':2,
    'c':1,
    'd':3,
    'e':2
}
ans = {
    val: [key for key, v in data.items() if v == val]
    for val in set(data.values())
}
print(ans)
```

8.  Given a list of student dictionaries like [{'name': 'Alice', 'marks': 85}, {'name': 'Bob', 'marks': 30}, ...], 
use a dictionary comprehension to build a new dict containing only students who passed (marks >= 
40), with names converted to uppercase as keys. 

```python
students = {"Omkar": 85, "Amit": 32, "Sneha": 91, "Riya": 38}

ans={name.upper():marks for name,marks in students.items() if marks>=40 }
print(ans)
```

10.  Use a nested list comprehension to build a 10x10 multiplication table as a list of lists. Then use a 
dictionary comprehension to convert that table into a dict mapping each row number (1-10) to its 
corresponding row (list).

```python
table = [[row * col for col in range(1, 11)] for row in range(1, 11)]
print(table)

ans = {row_num: row for row_num, row in enumerate(table, start=1)}

print(ans)
```
