4.  Given a dictionary mapping student names to letter grades, write code using .items() to print only the 
students who received an 'A'. 

```python
students = {
    "Omkar": "A",
    "Rahul": "B",
    "Sneha": "A",
    "Amit": "C"
}

for key,val in students.items():
    if val=='A':
        print(key)
```

6.  Given two sets, perform and print the union, intersection, and difference between them, clearly 
labeling each result. 

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

union = set1 | set2
inter = set1 & set2
diff = set1 - set2

print("Union:", union)
print("Intersection", inter)
print("Difference:", diff)
```

7.  Given a list of tuples in the form (name, score), sort the list so scores appear in descending order, and

8.  Write a program that counts the frequency of each word in a given paragraph (store counts in a 
dictionary), then prints the 3 most frequent words along with their counts. 


```python
paragraph = """Python is a popular programming language.
Python is easy to learn and Python is powerful.
Many developers use Python for data science.
Data science is growing and Python is widely used."""

words=paragraph.lower().split()
ans={}
for word in words:
    if word in ans:
        ans[word]+=1
    else:
        ans[word]=1
            
print(ans)    

sorted_ans=sorted(ans.items(),key=lambda x:x[1],reverse=True) 

print(sorted_ans[0:3])

```

9.  Given three separate lists of numbers (A, B, C), use set operations (no loops) to find the numbers that

10.  Design a simple contact book using a dictionary of dictionaries, e.g., {'Alice': {'phone': '...', 'email': 
'...'}}. Implement functions to add a new contact, update an existing contact's field, and delete a 
contact — then demonstrate all three operations. 


```python
contacts = {
    "Alice": {
        "phone": "9876543210",
        "email": "alice@gmail.com"
    }
}


# Add a new contact
def addNewContact(name, phone, email):
    contacts[name] = {
        "phone": phone,
        "email": email
    }


# Update an existing contact
def updateExisting(name, field, value):
    if name in contacts:
        contacts[name][field] = value
    else:
        print("Contact not found!")


# Delete a contact
def deleteContact(name):
    if name in contacts:
        contacts.pop(name)
    else:
        print("Contact not found!")


# Demonstrating all three operations
addNewContact("Bob", "9876543210", "bob@gmail.com")

updateExisting("Alice", "phone", "1234567890")

updateExisting("Alice", "email", "newalice@gmail.com")

deleteContact("Bob")

print(contacts)
```
