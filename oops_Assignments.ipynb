1. Library System
1.	Create class Book → attributes: title, author, is_issued
2.	Create class Member → name, member_id
3.	Create class Library → list of books
4.	Add method to add books
5.	Add method to issue book (change status)
6.	Add method to return book
7.	Print all available books


```python
class Book:
    def __init__(self,title,author):
        self.title=title
        self.author=author
        self.is_issued = False
class Member:
    def __init__(self,name,member_id):
        self.name=name
        self.member_id=member_id
class Library:
    def __init__(self):
        self.listB=[]    

    def addBook(self,book):
        self.listB.append(book)

    def display(self):
        for b in self.listB:
           if not b.is_issued:
             print(b.title,b.author)

    def issueBook(self,title):
        for b in self.listB:
            if title==b.title:
                if b.is_issued==False:
                    b.is_issued=True
                    print(f"The {b.title} is issued")

    def return_book(self,title):
        for b in self.listB:
            if title==b.title:
                if b.is_issued==True:
                    b.is_issued=False


b1=Book("Java","Dr ABC")   
b2=Book("CPP","Dr XYZ")   
l1=Library()
l1.addBook(b1)
l1.addBook(b2)
l1.display()
l1.issueBook("Java")
l1.return_book("Java")
l1.display()

```

2. Shopping Cart
1.	Create Product → name, price
2.	Create Cart → list of products
3.	Add method to add product
4.	Add method to remove product
5.	Add method to calculate total
6.	Display all items


```python
class Product:
    def __init__(self,name,price):
        self.name=name
        self.price=price
class Cart:
    def __init__(self):
        self.products=[]

    def addProduct(self,product):
        self.products.append(product)

    def removeProduct(self,product):
        if product in self.products:
            self.products.remove(product)

    def calculate(self):
        prices=[p.price for p in self.products]   
        total=sum(prices) 
        print(f"Total : {total}")

    def display(self):
        for p in self.products:
            print(f"Name : {p.name} Price: {p.price}")

p1=Product("Pen",300)
p2=Product("Charger",1000)            
c1=Cart()
c1.addProduct(p1)
c1.display()
c1.addProduct(p2)
c1.display()
c1.calculate()
c1.removeProduct(p1)
c1.display()
c1.calculate()


```

3. ATM Machine
1.	Create class ATM
2.	Add private attribute __balance
3.	Add method deposit(amount)
4.	Add method withdraw(amount) (check balance)
5.	Add method check_balance()
6.	Add PIN validation before operations


```python
class ATM:
    def __init__(self,balance,pin):
        self.__balance=balance
        self.pin=pin

    def deposit(self,amt,pin):
        if self.pin==pin:
            if amt<=0:
                print("Invalid Amount")
            else:
                self.__balance+=amt
                print(f"Amount Deposited: {amt}")
        else:
            print("Enter Valid Pin")

    def withdraw(self,amt,pin):
        if self.pin==pin:
            if amt>self.__balance:
                print("Amount cant be greater than balance")
            else:
                self.__balance-=amt     
                print("Amount withdrawed")
        else:
            print("Enter Valid pin")

    def checkBalance(self,pin):
        if self.pin==pin:
            print(f"Balance:{self.__balance}")

a1=ATM(50000,11)
a1.checkBalance(11)
a1.deposit(10000,11)
a1.checkBalance(11)
a1.withdraw(5000,11)
a1.checkBalance(11)
a1.checkBalance(12)
```

4. Student Report System
1.	Create Student class
2.	Store marks in list
3.	Add method calculate average
4.	Add method assign grade (A/B/C)
5.	Print report


```python
class Student:
    def __init__(self,name):
        self.name=name
        self.marklst=[]
        self.grade='F'

    def addmarks(self,mark):
        self.marklst.append(mark)

    
    def calculateavg(self):
        l=len(self.marklst)
        s=sum(self.marklst)
        if s<=0:
            print("Marks sum cant be zero")
        else:
            avr=s/l
            return avr
        
    def assignGrade(self):
        avg=self.calculateavg()
        if avg >= 80:
          self.grade='A'
        elif avg >= 60:
         self.grade="B"
        else:
            self.grade="C"

    def report(self):
        print(f"Name :{self.name}")
        print(f"Grade: {self.grade}")
        print("Marks: ")
        for mark in self.marklst:
            print(mark)


s1=Student("Omkar")
s1.addmarks(80)
s1.addmarks(90)
s1.addmarks(75)
s1.addmarks(60)
s1.assignGrade()
s1.report()
```

5. Rectangle vs Square
1.	Create Rectangle class → length, width
2.	Add area() and perimeter()
3.	Create Square class inheriting Rectangle
4.	Override constructor
5.	Test both


```python
class Rectangle:
    def __init__(self,l,w):
        self.l=l
        self.w=w
    def area(self):
        ar=self.l*self.w
        return ar
    def perimeter(self):
        pr=2*(self.l+self.w)
        return pr
class Square(Rectangle):
    def __init__(self,side):
        self.side=side
        self.side=side
    def area(self):
            ar=self.side**2
            return ar
    def perimeter(self):
            pr=4*self.side
            return pr

r1 = Rectangle(10, 5)
print(r1.area())     
print(r1.perimeter()) 
s1 = Square(5)
print(s1.area())      
print(s1.perimeter())   
```

10. Shape Polymorphism
1.	Create base class Shape
2.	Add method area()
3.	Create Circle, Triangle
4.	Override area()
5.	Call using same interface


```python
import math
class Shape:
    def area(self):
        pass
class Circle(Shape):
    def __init__(self,radius):
        self.radius=radius
    def area(self):
        a=math.pi*(self.radius**2)
        return a
class Traingle(Shape):
    def __init__(self,base,height):
        self.base=base
        self.height=height
    def area(self):
        a=0.5*self.base*self.height
        return a
c1=Circle(4)
print(c1.area())
t1=Traingle(10,5)
print(t1.area())
```

13. File System
1.	Create File class
2.	Create Folder class
3.	Folder holds list of files
4.	Add/remove files
5.	Display structure


```python
class File:
    def __init__(self,filename,filesize):
        self.filename=filename
        self.filesize=filesize

class Folder:
    def __init__(self,name):
        self.filelist=[]
        self.name=name

    def addfile(self,file):
        self.filelist.append(file)

    def removefile(self,file):
        if file in self.filelist:
            for f in self.filelist:
                if f==file:
                    self.filelist.remove(file)
        else:
            print("The file which you want to delete is not in the folder")

    def display(self):
        print(f"Folder Name:{self.name}")
        for f in self.filelist:
            print(f.filename,"-",f.filesize)
f1=File("main.py","10 KB")
f2=File("notes.txt","20kb")
fo1=Folder("ClassNotes")
fo1.addfile(f1)
fo1.addfile(f2)
fo1.display()
print("After")
fo1.removefile(f2)
fo1.display()


```

14. Quiz App

```python
class Question:
    def __init__(self,question,options,ans):
        self.question=question
        self.options=options
        self.ans=ans
        self.givenans=""

class Quiz:
    def __init__(self):
        self.listofque=[]

    def addque(self,que):
        self.listofque.append(que)

    def askqueandtakeans(self):
        for q in self.listofque:
            print(q.question)
            print(q.options)
            print("Give your option")
            given=int(input("Enter just option"))
            q.givenans=given
    def finalscore(self):
        ctr=0
        for q in self.listofque:
            if q.ans==q.givenans:
                ctr+=1
        print(f"Final Score:{ctr}/{len(self.listofque)}")

q1=Question("What is the capital of India?","1. Mumbai 2. Delhi3. Pune4. Chennai",2)
q2=Question("how many days in a yeaar","1.243 2.231 3.365 ,4.365",4)
qz=Quiz()
qz.addque(q1)
qz.addque(q2)
qz.askqueandtakeans()
qz.finalscore()

    
```

16. Chat Application
1.	Create User class
2.	Create Message class
3.	Store messages in list
4.	Add send/receive methods
5.	Display chat history


```python

class Message:
    def __init__(self,sender,receiver,content):
        self.sender=sender
        self.receiver=receiver
        self.content=content
class User:
    def __init__(self,username):
        self.username=username
        self.classhistory=[]

    def sendmessage(self,receiver,content):
        msgobj=Message(self,receiver,content)
        self.classhistory.append(msgobj)
        receiver.classhistory.append(msgobj)

    def displaychat(self):
        for msg in self.classhistory:
            print(msg.sender.username, ":", msg.content)
u1 = User("Omkar")
u2 = User("Rahul")

u1.sendmessage(u2, "Hey Rahul!")
u2.sendmessage(u1, "Hey Omkar!")
u1.sendmessage(u2, "How are you?")

u1.displaychat()
```
